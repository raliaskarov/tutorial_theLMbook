```python
from sklearn.feature_extraction.text import CountVectorizer # Text vectorization utility
from sklearn.linear_model import LogisticRegression         # Logistic regression model
from sklearn.metrics import accuracy_score                  # For model evaluation
```

```python
vectorizer = CountVectorizer(max_features=20, binary=True)
```

```python
X_train_text = ["I feel happy today", "I am very sad"]
X_test_text = ["happy but nervous"]

X_train = vectorizer.fit_transform(X_train_text)
# Vocabulary might contain: feel, happy, today, very, sad

X_test = vectorizer.transform(X_test_text)
# "happy" gets a feature value.
# "nervous" was never in the training vocabulary, so it is ignored.
```
