# TF IDF Model

```python
# vectorizer_demo.ipynb

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

# A small training corpus.
X_train_text = [
    "I feel happy and excited today",
    "This is a wonderful and joyful day",
    "I am sad and disappointed",
    "I feel angry about what happened",
    "I am nervous and worried about the exam",
]

# New, unseen documents: do not fit the vectorizer on these.
X_test_text = [
    "I am happy but a little nervous",
    "This situation makes me sad",
    "I feel excited about the project",
]

# Create the vectorizer.
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english"
)

# 1. Fit on TRAINING text and transform it into a TF-IDF matrix.
# This learns the vocabulary and IDF statistics.
X_train_vectors = vectorizer.fit_transform(X_train_text)

# 2. Transform TEST text using the vocabulary already learned above.
# Do NOT use fit_transform here.
# Do NOT call vectorizer._transform(...).
X_test_vectors = vectorizer.transform(X_test_text)

print("Vocabulary:")
print(vectorizer.vocabulary_)

print("\nTraining matrix shape:")
print(X_train_vectors.shape)

print("\nTest matrix shape:")
print(X_test_vectors.shape)

# Display the matrices in a readable format.
feature_names = vectorizer.get_feature_names_out()

train_df = pd.DataFrame(
    X_train_vectors.toarray(),
    columns=feature_names,
    index=X_train_text
)

test_df = pd.DataFrame(
    X_test_vectors.toarray(),
    columns=feature_names,
    index=X_test_text
)

print("\nTraining TF-IDF features:")
display(train_df)

print("\nTest TF-IDF features:")
display(test_df)
```
