```python
# pip install -U transformers accelerate
# Load model directly
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen2.5-1.5B-Instruct")
model = AutoModelForCausalLM.from_pretrained("Qwen/Qwen2.5-1.5B-Instruct", device_map="auto")
```

```python
from IPython.display import Markdown, display
display(Markdown(generate("Explain CaCO3 for clinker production")))

def generate(user_prompt):
  messages = [
    {"role": "user", "content": user_prompt},
  ]
  inputs = tokenizer.apply_chat_template(
    messages,
    add_generation_prompt=True,
    tokenize=True,
    return_dict=True,
    return_tensors="pt",
  ).to(model.device)

  outputs = model.generate(**inputs, max_new_tokens=256)
  response = tokenizer.decode(outputs[0][inputs["input_ids"].shape[-1]:])
  return response

def generate_and_print(user_prompt):
  return display(Markdown(generate(user_prompt)))
```

```python

user_prompts = ["Who are you?",
                "How much money will I have in back account in 3 years if I deposit USD 100 at 5% interest?",
                 "I have 3 apples, my friend eats one and I give 2 to my sister, how many apples do I have now?",
                "Solve the equation 6x + 4 = 40"
]

for user_prompt in user_prompts:
  print("-------------------------------------")
  print("Prompt:\n", user_prompt)
  print("Response:")
  generate_and_print(user_prompt)
```
